```python
#enumerate
using this function we can define an index of a given items from the iterable, by default your index will starts from 0 to n -1, if you want change it you can do that.

enumerate will give you the result in the form of tuple.
```

```python
for pair in list(enumerate(lst)):
    if pair[-1] == 11:
        print(pair, end = " ")
```

```python
for pair in list(enumerate(lst)):
    if pair[-1]==11:
        print(pair[0],end=" ")  #all index value are 11
```
