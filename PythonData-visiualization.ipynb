```python
rs=df.groupby("Sub-Category")['Sales'].sum()
```

```python
pd.set_option('display.precision',2)
```

```python
rs
```

```python
rs.plot(figsize=(10,2))
plt.show ()      # plot -> line chart
```

```python
rs.plot(figsize=(8,3), kind='bar')
plt.show()
```

```python
rs.plot(figsize=(3,5), kind='barh')
plt.show()
```

```python
rs = rs.reset_index()
rs.head()
```

```python
plt.figure(figsize=(20,5))
plt.bar(rs['Sub-Category'], rs['Sales'], color = 'r', label = 'mySales')
plt.xlabel('Sub-Category')
plt.ylabel('Sales')
plt.legend()
plt.title("Sub-Category wise Sales")
plt.show()
```

```python
plt.figure(figsize=(30,5))
plt.bar(rs['Sub-Category'],rs['Sales'],color='g', label='mySales')

plt.show()
```

```python
# rotate x ticks
plt.figure(figsize=(20,5))
plt.bar(rs['Sub-Category'], rs['Sales'], color = 'c', label = 'mySales')
plt.xlabel('Sub-Category')
plt.ylabel('Sales')
plt.xticks(rotation = 90) #ye add kr denge for rotation
plt.legend()
plt.title("Sub-Category wise Sales")
plt.show()
```

```python
plt.figure(figsize=(4,5))
plt.barh(rs['Sub-Category'], rs['Sales'], color = 'r', label = 'mySales')
plt.xlabel('Sub-Category')
plt.ylabel('Sales')
plt.xticks(rotation = 60)
plt.legend()
plt.title("Sub-Category wise Sales")
plt.show()
```

```python
ig, ax = plt.subplots(figsize = (12,4))
bar = ax.bar(rs['Sub-Category'], rs.Sales)
ax.bar_label(bar, label_type = 'edge', fontsize = 8, color = 'blue')
plt.xticks(rotation = 90)
plt.legend()
plt.title("Sub-Category wise Sales")
plt.show()
```

```python
name=['Abhishek','Anuradha','Mohit']
sales=[380,480,780]
```

```python
zip(name,sales)
```

```python
list(zip(name,sales))
```

```python
for n,s in zip(name,sales):
    print(n,s)
```

```python
lbl = [f"{n}, {s}" for n, s in zip(name,sales)] # list comprehension : DDR 
```

```python
plt.figure(figsize=(5,3))
plt.pie(x=sales,labels=lbl)
plt.title("sales dstribution", fontsize=7)
plt.show()
```

```python
plt.figure(figsize=(5,3))
plt.pie(x=sales,labels=lbl,wedgeprops={"edgecolor":"black","linewidth":0.5},
        autopct="%1.2f%%",
        colors={"red","blue","green"},
        explode=[0.02,0.02,0.02],
        startangle=90
       )
plt.title("sales dstribution", fontsize=9)
plt.show()
```

```python
rs=pd.pivot_table(df,index="Sub-Category", values=["Quantity","Sales","Profit"],
                  aggfunc=sum).reset_index()
rs.head(2)
```

```python
rs.to_excel("sub_category_sales.xlsx",index=False)
```

```python
plt.figure(figsize=(10,3))
plt.bar(rs['Sub-Category'],rs['Sales'],color="#4281f5",label="Sales")
plt.plot(rs['Sub-Category'], rs['Profit'],"*--r", label = "Profit")
plt.plot(rs['Sub-Category'], rs['Quantity'],"--b", label = "Quantity")
plt.legend()
plt.xticks(rotation=90)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 3))

ax.bar(rs['Sub-Category'], rs['Sales'],color="#526494", label="Sales")
ax.set_ylabel('Sales', color = 'black')
ax.set_xticklabels(rs['Sub-Category'], rotation = 90)

ax1 = ax.twinx()
ax1.plot(rs['Sub-Category'], rs['Profit'],"*--b", label="Profit")
ax1.plot(rs['Sub-Category'], rs['Quantity'],"x-g", label="Quantity")
# ax.set_yticklabels(rs.Profit, rotation = 10)

plt.legend()
plt.show()
```
