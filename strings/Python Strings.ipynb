## Partition Function

Using this function we can partition the data based on given char, this function always return an output in the form of tuple and the tuple contains at least or at most only 3 items, this function partition the data from the left side.

If the given char is not part of the given string still this function return 3 items in the tuple as follows:

1. before the separator
2. separator itself
3. after the separator

If separator not then it will give you whole string with 2 empty string.

```python
email = "myprogrammingisfun@gmail.com"
```

```python
email.split("@")
```

```python
# how partition works
```

```python
email.partition("@")
```

```python
email.partition("")
```
